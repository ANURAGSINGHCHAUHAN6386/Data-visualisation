# series method
* item method

```python
import numpy as np
s=pd.Series(np.arange(1,10,2),index=['a','b','c','d','e'],name='Odd')
print(s,type(s))
```

```python
# iteration over a series iterates over the value
for value in  s:
    print(value)
    print("_"*45)
    for item in s.items():
        print(item)
```

```python
for idx ,value in s.items():
    print(idx,value)
```

```python
for idx in s.keys():
    print(idx,"=>",s[idx]) #idx ,value =("rahul",22)
```

# Reset Index
* The reset_index()function is used to generate a new datframe  or series with the index reset
* This is used when the index need to be treated as  column or when the index is meanigless and needs to be resert the default before another operation
* Series.reset_index (self,level=None ,drop=False,name=None ,inplace=False) 

```python
#create a series
s=pd.Series([2,3,4,5],
                 name='f1',
                index=pd.Index(['p','q','r','s'],name="idx"))
print(s)
```

```python
x=s.reset_index()
print(type(x))
print(x)
```

```python
s.reset_index(name="value")
print(s)
```

```python
s.reset_index(drop=True)
print(x,(type(x)))
```

```python
# implace =true reflect the change in the original series instead of temperory changes
s.reset_index(inplace=True,drop=True)
print(s)
```
