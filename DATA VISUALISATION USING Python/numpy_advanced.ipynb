# aggregations (reduce method)
* if you like to reduce an array with a particuler operetion we can use the reduce method of any ufunc
* a reduce repeatetly appield a givin operation to the element of an array until only a single result remains.

```python
import numpy as np
x=np.arange(1,11)
print(x)
np.add.reduce(x)
print(np.sum(x)) # function to compute the result
print(np.prod(x))
print(np.cumsum(x))
print(np.cumprod(x))
print(np.add.accumulate(x))

```

# Arithmetic with NumPy Arrays

| Operation | NumPy Function | Result |
|---|---|---|
| Addition | `np.add(a,b)` | `[6, 8, 10, 12]` |
| Subtraction | `np.subtract(a,b)` | `[-4, -4, -4, -4]` |
| Multiplication | `np.multiply(a,b)` | `[5, 12, 21, 32]` |
| Division | `np.divide(a,b)` | `[0.2, 0.333, 0.429, 0.5]` |
| Power | `np.power(a,2)` | `[1, 4, 9, 16]` |
| Modulus | `np.mod(a,2)` | `[1, 0, 1, 0]` |
| Negative | `np.negative(a)` | `[-1, -2, -3, -4]` |

```python
import numpy as np

x = np.array([1, 2, 3, np.nan])
y = np.array([9, 7, 6, 5])

print(np.add.reduce(x))
print(np.sum(x))
print(np.nansum(x))

print(np.add.reduce(y))
print(np.sum(y))

print(np.prod(x))
print(np.prod(y))
```

## Searching Array
* you can search an array for a certain value and return the indexes that the get  mathcg
to search an arcray , use the where () method

```python
import numpy as np
x=np.random.randint(1,10,size=10)
print(x)
print(np.where(x>4))
print(np.where(x==4))
```
