## Vectorization 
* Vectorization means performing operation on entire array at once instead of using python loops.
* it is fast as numpy used optiized C code internally
* works on same shape array.

```python
# vectorization
#with list you do oparation element by ellemnt(slow).
# Numpy allow element-wise opration without explicit loops
## Vectorization 
import numpy as np
arr=np.arange(100000)
lst=[x for x in range (100000)]
# python list
start =time.time()
lst2=[x*2 for x in lst]
end=time.time()
print(end-start)
#numpy array
start1=time.time()
arr2=arr*2 #vectoraized
end1=time.time()
print(end-start1)


```

## Broadcasting 
* Opearations btw differant shape arrays.
* Numpy automatically expands the smaller array (without copying ) so shape become compatible


```python
import numpy as np
a=np.array([1,2,3])
b=5 #scalar
#here, 5 is broadcast[5,5,5,]
print(a+b)
```

```python
import numpy as np
a=np.array([1,2,3])
b=5 #scalar
#here, 5 is broadcast[5,5,5,]
print(a**b)
```

```python
import numpy as np

arr = np.array([1, 2, 3])       # shape (3,)
vec = np.array([[1], [2], [3]]) # shape (3,1)

print(arr + vec)
```

# indexing and slicing
* context of ndarray object can be accessed and modified by indexing or slicing


```python
import numpy as np
x=np.array([10,20,30,40])
print(x[1:4])
print(x[1:])
```

```python
import numpy as np
y = np.array([[10, 20, 30, 40], [12, 234, 53, 64], [32, 35, 565, 75]])
print(y[1:])
print(y[:2])
```

```python
import numpy as np
# all item from colom 1 onward
y = np.array([
    [22, 2, 34],
    [32, 45, 6],
    [43, 56, 6],
    [1, 1, 1]
])
print(y[0:2,1:])
print(y[:,1:])
```

```python
# sliccing returns a view
# All updation is performed innt he original array
import numpy as np
arr=np.array([1,2,3,4])
s=arr[1:3]
s[0]=100
print(arr) #([1 100 3 4])
print(s)
print(s.base) # this will return the original array
```

```python
#fancy indexing retrus a copy
import numpy as np
arr=np.array([1,2,3,4])
f=arr[[1,2]]  #1 and 2 is indexing number
print(f)
f[0]=500 
print(arr) #unchanged
print(f)
print(f.base) #iska koi kud ka original 
```
