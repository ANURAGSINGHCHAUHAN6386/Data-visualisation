```python
aqi = int(input("Enter AQI: "))

if aqi <= 50:
    print("Air Quality: Good")
    print("Purifier: OFF")

elif aqi <= 100:
    print("Air Quality: Moderate")
    print("Purifier: LOW SPEED")

elif aqi <= 200:
    print("Air Quality: Poor")
    print("Purifier: MEDIUM SPEED")

elif aqi <= 300:
    print("Air Quality: Very Poor")
    print("Purifier: HIGH SPEED")
    print("Alert: Wear a Mask")

else:
    print("Air Quality: Hazardous")
    print("Purifier: MAX SPEED")
    print("Emergency Alert Sent")
```

```python
year = int(input("Enter the year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

```python

#simple calculater
num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))
operation = input("Enter operation (+, -, *, /): ")

if operation == '+':
    result = num1 + num2

elif operation == '-':
    result = num1 - num2

elif operation == '*':
    result = num1 * num2

elif operation == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error! Division by zero"

else:
    result = "Invalid operation"

print("Result is:", result)
```
