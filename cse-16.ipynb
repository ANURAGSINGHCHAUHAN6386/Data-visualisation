```python
#. List
# list - collection of heterogeneous data type , ordered ,mutable-changeable , allowed deplication
# Ordered (items have a fixed position)
#  Mutable (items can be changed)
#  Allows duplicate values
#  Can store different data types

A = [1,4.2,None,3,1,4.2,6, 'Hello',6+6j,True]₹
print(A,type(A))
A[5] = "HI"
print(A[5])
print(A)
```

```python
#tuple
#  Ordered (elements have a fixed position)
#  Immutable (cannot be changed after creation)
#  Allows duplicate values
#  Can store different data types

A = (1,4.2,None,3,1,4.2,6, 'Hello',6+6j,True)
print(A,type(A))
# A[5] = "HI"
print(A[5])
print(A)
```

```python
# # SETS
#  Unordered (no indexing)
#  Mutable (can add or remove elements)
#  Stores only unique values (duplicates are removed)
#  Can contain different data types (if hashable)

A = {1,4.2,None,3,1,4.2,6, 'Hello',6+6j,True}
print(A,type(A))
# A[5] = "HI"
# print(A[5])
# print(A)
for i in A:
    print(i)
```

```python
##. Dictionary
#. Stores data as key : value pairs
# Mutable (can add, update, or delete items)
#  Keys must be unique
#  Values can be duplicated
#  Access values using keys
D = {1:"a", 2:"b", 3+3j:"c", "Hii":"d", 5.3:"e"}
print(D)
print(type(D))
```

what is shallow copy
-->A shallow copy creates a new object, but does not create copies of nested (inner) objects. Instead, it copies the references to those nested objects.
Top-level object → New
Nested objects → Shared

what is deep copy
-->A deep copy creates a new object and also creates copies of all nested objects.
Top-level object → New
Nested objects → New
